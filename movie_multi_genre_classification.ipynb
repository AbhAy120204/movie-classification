```python
import kagglehub

path = kagglehub.dataset_download("rajugc/imdb-movies-dataset-based-on-genre")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import re
```

```python
genre1 = pd.read_csv(f'{path}/action.csv')
genre2 = pd.read_csv(f'{path}/crime.csv')
genre3 = pd.read_csv(f'{path}/adventure.csv')
genre4 = pd.read_csv(f'{path}/thriller.csv')
genre5 = pd.read_csv(f'{path}/family.csv')
genre6 = pd.read_csv(f'{path}/mystery.csv')
genre7 = pd.read_csv(f'{path}/scifi.csv')
genre8 = pd.read_csv(f'{path}/history.csv')
genre9 = pd.read_csv(f'{path}/sports.csv')
genre10 = pd.read_csv(f'{path}/animation.csv')
genre11 = pd.read_csv(f'{path}/war.csv')
genre12 = pd.read_csv(f'{path}/biography.csv')
genre13 = pd.read_csv(f'{path}/horror.csv')
genre14 = pd.read_csv(f'{path}/fantasy.csv')
genre15 = pd.read_csv(f'{path}/romance.csv')
genre16 = pd.read_csv(f'{path}/film-noir.csv')
```

```python
df = pd.concat([genre1, genre2, genre3, genre4,
                genre5, genre6, genre7, genre8,
                genre9, genre10, genre11, genre12,
                genre13, genre14, genre15, genre16])

df = df.reset_index(drop=True)

df
```

```python
df['genre'] = df['genre'].apply(lambda x: [g.strip() for g in x.split(',')] if isinstance(x, str) else [])

df = df.groupby('movie_id').agg({
    'movie_name': 'first',
    'year': 'first',
    'certificate': 'first',
    'runtime': 'first',
    'genre': lambda x: sorted(set(g for sublist in x for g in sublist)),  # merge and deduplicate
    'rating': 'mean',
    'description': 'first',
    'director': 'first',
    'director_id': 'first',
    'star': 'first',
    'star_id': 'first',
    'votes': 'mean',
    'gross(in $)': 'mean'
}).reset_index()
```

```python
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
df['description'] = df['description'].apply(clean_text)
```

```python
all_genres = [genre for genres in df['genre'] for genre in genres]
all_genres[:3]
```

```python
unique_labels = sorted(set(genre for genres in df['genre'] for genre in genres))
print(len(unique_labels))
print(unique_labels)
```

```python
id2label = {idx:label for idx, label in enumerate(unique_labels)}
label2id = {label:idx for idx, label in enumerate(unique_labels)}
```

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer

MODEL = 'distilbert-base-uncased'

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL,
    problem_type = "multi_label_classification",
    num_labels=len(unique_labels),
    id2label=id2label,
    label2id=label2id
    )
```

```python
import numpy as np
import torch
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL)
```

```python
def preprocess_data(examples,text_col):
  one_hot_encoded_matrix = []
  text = examples[text_col]
  genres = examples["genre"]
  for genre in genres:
    g = genre
    one_hot_encoded_row = [1 if label in g else 0 for label in unique_labels]
    one_hot_encoded_matrix.append(one_hot_encoded_row)

  one_hot_encoded_matrix = [torch.tensor(o, dtype = torch.float32) for o in one_hot_encoded_matrix]

  encoding = tokenizer(text, truncation=True, max_length=256)

  encoding["labels"] = one_hot_encoded_matrix

  return encoding
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

# Multi-label binarization
mlb = MultiLabelBinarizer()
genre_encoded = pd.DataFrame(mlb.fit_transform(df['genre']), columns=mlb.classes_)

# Combine with the original DataFrame (drop old genre if needed)
df_encoded = pd.concat([df.drop(columns=['genre']), genre_encoded], axis=1)
```

```python
!pip install iterative-stratification
```

```python
from iterstrat.ml_stratifiers import MultilabelStratifiedKFold

# X: input features, y: multi-label genre matrix
X = df_encoded[['description']]
y = genre_encoded

# Stratified split
mskf = MultilabelStratifiedKFold(n_splits=5, shuffle=True, random_state=42)
train_idx, test_idx = next(mskf.split(X, y))

# Create train/test DataFrames
train_df = df.iloc[train_idx].reset_index(drop=True)
test_df = df.iloc[test_idx].reset_index(drop=True)
```

```python
from datasets import Dataset, DatasetDict

# Optional: Keep only relevant columns
train_dataset = Dataset.from_pandas(train_df[['genre', 'description']])
test_dataset = Dataset.from_pandas(test_df[['genre', 'description']])

# Create a DatasetDict for easy model usage
movie_dataset = DatasetDict({
    'train': train_dataset,
    'test': test_dataset
})
```

```python
description_encoded_dataset = movie_dataset.map(
    lambda x: preprocess_data(x, 'description'),
    batched=True,
    batch_size=128,
    remove_columns=['genre', 'description']
)
```

```python
from sklearn.metrics import f1_score, roc_auc_score, accuracy_score
from transformers import EvalPrediction
import torch
import numpy as np
from sklearn.metrics import f1_score, roc_auc_score, accuracy_score, jaccard_score
```

```python
def multi_label_metrics(predictions, labels, threshold=0.5):

    sigmoid = torch.nn.Sigmoid()

    probs = sigmoid(torch.Tensor(predictions))

    y_pred = np.zeros(probs.shape)
    y_pred[np.where(probs >= threshold)] = 1

    y_true = labels

    f1_micro_average = f1_score(y_true=y_true, y_pred=y_pred, average='micro')
    roc_auc = roc_auc_score(y_true, y_pred, average = 'micro')
    accuracy = accuracy_score(y_true, y_pred)
    jaccard = jaccard_score(y_true, y_pred, average='micro')

    metrics = {'f1': f1_micro_average,
              'roc_auc': roc_auc,
              'accuracy': accuracy,
              'jaccard': jaccard}

    return metrics

def compute_metrics(p: EvalPrediction):

    preds = p.predictions[0] if isinstance(p.predictions, tuple) else p.predictions

    result = multi_label_metrics(predictions=preds, labels=p.label_ids)

    return result
```

```python
from transformers import TrainingArguments, Trainer, DataCollatorWithPadding
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

```python
args = TrainingArguments(
    f"genre-prediction",
    eval_strategy = "epoch",
    save_strategy = "epoch",
    per_device_train_batch_size=16,
    gradient_accumulation_steps=4,
    per_device_eval_batch_size=64,
    num_train_epochs=10,
    logging_steps=10,
    load_best_model_at_end=True,
    metric_for_best_model='jaccard',
)

trainer = Trainer(
    model,
    args,
    train_dataset=description_encoded_dataset["train"],
    eval_dataset=description_encoded_dataset["test"],
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

trainer.evaluate()
```

```python
from transformers import pipeline, AutoModelForSequenceClassification, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(MODEL)

trained_model = AutoModelForSequenceClassification.from_pretrained(
    f"genre-prediction",
    problem_type = "multi_label_classification",
).eval()

classifier = pipeline('text-classification', model=trained_model, tokenizer=tokenizer, return_all_scores=True)
```
